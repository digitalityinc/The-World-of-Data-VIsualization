# using Pandas and Heatmap

```python
import pandas as pd
import numpy as np

missing_values = ["n/a", "na", "---", "D", "..", "-"]
hw1_df = pd.read_csv("https://raw.githubusercontent.com/huizhangky/CSE590/main/hw/hw1/USA.csv", encoding = "ISO-8859-1", na_values = missing_values)
```

```python
hw1_df
```

```python
hw1_heatmap_df = hw1_df.copy()

hw1_heatmap_df = hw1_heatmap_df.query("Country != 'Unknown' and Country != 'Total'")

hw1_heatmap_coutry = hw1_heatmap_df['Country'].tolist()

hw1_heatmap_df.drop(columns = ['Type', 'Coverage', 'Country', 'AREA', 'Continent', 'REG', 'Region', 'DEV', 'DevName'], axis=1, inplace=True)
```

```python
hw1_heatmap_df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize = (30, 30))
sns.heatmap(hw1_heatmap_df, cmap="Greens", linewidths = 0.10)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize = (30, 100))
sns.heatmap(hw1_heatmap_df, cmap="Greens", linewidths = 0.10, yticklabels=hw1_heatmap_coutry)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
plt.figure(figsize = (30, 100))
ax = sns.heatmap(hw1_heatmap_df, cmap="Greens", linewidths = 0.10, yticklabels=hw1_heatmap_coutry)
ax.set_yticklabels(ax.get_yticklabels(), rotation=10)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
fig, ax = plt.subplots(figsize = (30, 100))
sns.heatmap(hw1_heatmap_df,yticklabels=hw1_heatmap_coutry, cbar_kws={"shrink": 0.1}, linewidths=0.1, linecolor="grey", cmap="Greens")
```

# Styling for your Pandas DataFrame

**Highlight values in your dataframe with a condition**

```python
import pandas as pd
import numpy as np

def color_negative_red(val):
    color = 'red' if val < 0 else 'black'
    return 'color: %s' % color

df = pd.DataFrame(dict(col_1=[1.53,-2.5,3.53], 
                       col_2=[-4.1,5.9,0])
                 )
df.style.applymap(color_negative_red)
```

**Hovering effect**

```python
df = pd.DataFrame(np.random.randn(5, 3))
df.style.set_table_styles(
    [{'selector': 'tr:hover',
      'props': [('background-color', 'yellow')]}]
)
```

**use CSS to change the appearance of the table**

```python
df = pd.DataFrame(
dict(departure=['SFO', 'SFO', 'LAX', 'LAX', 'JFK', 'SFO'],
     arrival=['ORD', 'DFW', 'DFW', 'ATL', 'ATL', 'ORD'],
     airlines=['Delta','JetBlue','Delta','AA','SouthWest',  
               'Delta']),
columns=['airlines', 'departure','arrival'])
df.style.set_table_styles(
[{'selector': 'tr:nth-of-type(odd)',
  'props': [('background', '#eee')]}, 
 {'selector': 'tr:nth-of-type(even)',
  'props': [('background', 'white')]},
 {'selector': 'th',
  'props': [('background', '#606060'), 
            ('color', 'white'),
            ('font-family', 'verdana')]},
 {'selector': 'td',
  'props': [('font-family', 'verdana')]},
]
).hide_index()
```
